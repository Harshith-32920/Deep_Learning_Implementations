```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input,SimpleRNN,Embedding, LSTM, Dense
```

```python
# Shuffled dataset of complete, short sentences for binary sentiment classification
sentences = [
    "She solved the difficult math problem quickly.",
    "We had a terrible time at the park.",
    "The food at this restaurant is delicious.",
    "I really hated reading that new book.",
    "The new software update works flawlessly now.",
    "This ugly painting ruins the entire hallway.",
    "The food at this restaurant is disgusting.",
    "He always greets everyone with a smile.",
    "She failed to solve the math problem.",
    "I am very proud of your hard work.",
    "My team lost the final championship game.",
    "I really enjoyed reading that new book.",
    "We had a wonderful time at the park.",
    "I am very disappointed by your laziness.",
    "My team won the final championship game.",
    "He always ignores everyone in the office.",
    "The bright morning sun warmed the room.",
    "The new software update crashes completely now.",
    "This beautiful painting brightens the entire hallway.",
    "The dark storm clouds ruined the afternoon."
]

# Labels matched to the shuffled sentences: 1 for positive, 0 for negative
labels = [1]*10 + [0]*10
labels=np.array(labels)
```

```python
vocab_size=2000
tokenizer=Tokenizer(num_words=vocab_size,oov_token="<OOV>")
#break th etexts into tokens - embeddings -OHE
tokenizer.fit_on_texts(sentences)
#convert texts to sequencs
sequences=tokenizer.texts_to_sequences(sentences)
maxlen=max(len(s) for s in sequences)
X=pad_sequences(sequences,maxlen=maxlen,padding="post")
y=labels
```

```python
#len of the longest sentence in the sentences array
maxlen
```

```python
#tokens of the last sentence sin the sentences array
X[19]
```

Model Creation

```python
# creating embeddings of each word
embed_dim=16
#number of units in each embedding\
rnn_units=8
```

```python
inp=Input(shape=(maxlen,),dtype='int32',name="input")
x=Embedding(input_dim=vocab_size,output_dim=embed_dim,mask_zero=True,name='embed')(inp)
rnn= SimpleRNN(units=rnn_units,return_sequences=False,name='simple_rnn')
x_last=rnn(x)
out=Dense(1,activation='sigmoid',name='out')(x_last)
model=Model(inputs=inp,outputs=out)
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X,y,epochs=25,batch_size=8,verbose=1)
```

### Model Evaluation and Prediction

First, let's evaluate the model on the training data. This will give us an idea of how well it learned the patterns in our existing sentences.

```python
loss, accuracy = model.evaluate(X, y, verbose=0)
print(f"Model Loss on training data: {loss:.4f}")
print(f"Model Accuracy on training data: {accuracy:.4f}")
```

Now, let's test the model with some new sentences to predict their sentiment. We'll use the same tokenizer and padding logic.

```python
new_sentences = [
    "I love this movie, it's fantastic!",
    "What a terrible experience, completely ruined my day.",
    "The weather is beautiful today.",
    "I am so bored right now."
]

# Tokenize and pad the new sentences
new_sequences = tokenizer.texts_to_sequences(new_sentences)
new_X = pad_sequences(new_sequences, maxlen=maxlen, padding="post")

# Make predictions
predictions = model.predict(new_X)

# Display predictions
for i, sentence in enumerate(new_sentences):
    sentiment = "Positive" if predictions[i][0] > 0.5 else "Negative"
    print(f"Sentence: '{sentence}' -> Predicted Sentiment: {sentiment} (Score: {predictions[i][0]:.2f})")
```

```python
print("-" * 80)
print(f"{'Sentence':<40} {'Sentiment':<15} {'Score':<10}")
print("-" * 80)

for i, sentence in enumerate(new_sentences):
    score = predictions[i][0]
    sentiment = "Positive" if score > 0.5 else "Negative"

    print(f"{sentence:<40} {sentiment:<15} {score:<10.2f}")

print("-" * 80)
```
